```python
import random
from datetime import datetime
from typing import List

# Entities reused from previous design
class Badge:
    def __init__(self, badge_id: str, name: str, description: str):
        self.badge_id = badge_id
        self.name = name
        self.description = description

class Review:
    def __init__(self, review_id: str, reviewer_id: str, content: str, target_entity: str):
        self.review_id = review_id
        self.reviewer_id = reviewer_id
        self.content = content
        self.target_entity = target_entity
        self.timestamp = datetime.utcnow()
        self.is_verified = False
        self.moderated_by = None

class Endorsement:
    def __init__(self, endorsement_id: str, endorser_id: str, endorsee_id: str, stake_amount: float):
        self.endorsement_id = endorsement_id
        self.endorser_id = endorser_id
        self.endorsee_id = endorsee_id
        self.stake_amount = stake_amount
        self.timestamp = datetime.utcnow()

class User:
    def __init__(self, user_id: str, is_good: bool):
        self.user_id = user_id
        self.is_good = is_good
        self.reputation_score = 0
        self.reviews: List[Review] = []
        self.endorsements_given: List[Endorsement] = []
        self.endorsements_received: List[Endorsement] = []
        self.badges: List[Badge] = []

    def submit_review(self):
        review_content = "Great product!" if self.is_good else "Spam content"
        review = Review(
            review_id=f"rev_{self.user_id}_{len(self.reviews)}",
            reviewer_id=self.user_id,
            content=review_content,
            target_entity="product_001"
        )
        self.reviews.append(review)
        if self.is_good:
            self.reputation_score += 5
        else:
            self.reputation_score -= 3

    def endorse_user(self, target_user):
        stake = 10.0 if self.is_good else 1.0
        endorsement = Endorsement(
            endorsement_id=f"endorse_{self.user_id}_{target_user.user_id}",
            endorser_id=self.user_id,
            endorsee_id=target_user.user_id,
            stake_amount=stake
        )
        self.endorsements_given.append(endorsement)
        target_user.endorsements_received.append(endorsement)
        if self.is_good and target_user.is_good:
            target_user.reputation_score += 2
        elif not self.is_good:
            target_user.reputation_score -= 1

# Generate users
users = []
for i in range(50):
    is_good = random.random() < 0.7  # 70% chance of being a good user
    user_type = "GoodUser" if is_good else "BadUser"
    user = User(user_id=f"{user_type}_{i}", is_good=is_good)
    users.append(user)

# Simulate behaviors
for user in users:
    for _ in range(random.randint(1, 5)):
        user.submit_review()
    targets = random.sample(users, k=random.randint(1, 3))
    for target in targets:
        if target != user:
            user.endorse_user(target)

# Collect results
user_summary = [{
    "user_id": user.user_id,
    "is_good": user.is_good,
    "reputation_score": user.reputation_score,
    "reviews": len(user.reviews),
    "endorsements_given": len(user.endorsements_given),
    "endorsements_received": len(user.endorsements_received)
} for user in users]

user_summary[:50]  # Show summary for first 5 users as sample
```
